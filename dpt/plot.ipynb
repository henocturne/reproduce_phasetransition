```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


def load_data(path):
    pd.read_csv(path,delimiter=' ',header=0)
    

```

```python

'''
====================================================================================================
多图设置
====================================================================================================
'''
fig1, ax1 = plt.subplots(
    nrows=1,
    ncols=1,
    figsize=(10, 6)
)

data = []
label = []
axis_name = ["x","y"]
data_x,data_y = data
data_x = data[0]
data_y = data[1:]


#坐标范围和间隔
x_tick_bound = []
y_tick_bound = []
x_min,x_max,x_interval = x_tick_bound
y_min,y_max,y_interval = y_tick_bound

#设置坐标轴和刻度字体大小
ax1.set_xlabel(axis_name[0],fontsize=28)
ax1.set_ylabel(axis_name[1],fontsize=28)
ax1.tick_params(axis='x', labelsize=28)
ax1.tick_params(axis='y', labelsize=28)

#设置坐标轴线宽
ax1.set_xticks(np.arange(x_min,x_max+1,x_interval))
ax1.set_yticks(np.arange(y_min,y_max+1,y_interval))

#边框线宽
ax1.spines['bottom'].set_linewidth(2)
ax1.spines['left'].set_linewidth(2)
ax1.spines['top'].set_linewidth(2)
ax1.spines['right'].set_linewidth(2)

#确定原点，网格
plt.xlim(left=0)
plt.ylim(bottom=0)
ax1.grid(True, linestyle='--', linewidth=0.8, alpha=0.7)

#标注图片序号
ax1.text(
    #order 'a'
    0.0, 1.1, 'a',
    transform=ax1.transAxes,
    fontsize=28,
    fontweight='bold',
    va='top', ha='left',
    clip_on=False 
)

plt.scatter(data_x, data_y
            ,marker = 'o'
            ,facecolors='none'   # 内部空心
            ,edgecolors='blue'   # 边框颜色
            ,alpha=0.7
            ,linewidths=2
            ,s=100
            ,label = 'points'
            )

#绘制曲线
ax1.plot(data_x,data_y[0],color="green", alpha=0.5,linestyle='-',label=label[0])
ax1.plot(data_x,data_y[1],color="green", alpha=0.5,linestyle='-',label=label[1])
ax1.legend(fontsize=28,loc="lower right",frameon=False)


plt.tight_layout() 
```
